```python
import unittest
def count_vowels(text: str) -> int:
    vowels = "aeiouyаеєиіїоуюя" 
    count = 0
    for char in text.lower():
        if char in vowels:
            count += 1
    return count

class TestCountVowels(unittest.TestCase):
    
    def test_normal_string(self):
        self.assertEqual(count_vowels("hello"), 2)
        
    def test_empty_string(self):
        self.assertEqual(count_vowels(""), 0)
        
    def test_numbers_string(self):
        self.assertEqual(count_vowels("12345!@#"), 0)
        
    def test_ukrainian_string(self):
        self.assertEqual(count_vowels("Тестування"), 4)

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

```python
class Figure:
    FIGURES = ["квадрат", "прямокутник", "трикутник"]

    def __init__(self, figure_type: str, length: int) -> None:
        assert length > 0, "Довжина має бути більшою за 0!"
        assert figure_type in self.FIGURES, "Невідомий тип фігури"
        self.type = figure_type
        self.length = length

    @property
    def get_figure_type(self) -> str:
        return self.type

    @property
    def get_figure_length(self) -> int:
        return self.length
    
    @property
    def get_angles(self) -> int:
        if self.type == "трикутник":
            return 3
        elif self.type in ["квадрат", "прямокутник"]:
            return 4

    
import unittest

class TestFigure(unittest.TestCase):
    
    def test_figure_length(self):
        fig = Figure("квадрат", 10)
        self.assertEqual(fig.get_figure_length, 10)
        
    def test_get_angles(self):
        square = Figure("квадрат", 5)
        rect = Figure("прямокутник", 5)
        triangle = Figure("трикутник", 5)
        
        self.assertEqual(square.get_angles, 4)
        self.assertEqual(rect.get_angles, 4)
        self.assertEqual(triangle.get_angles, 3)

    def test_invalid_length_and_type(self):
        with self.assertRaisesRegex(AssertionError, "Довжина має бути більшою за 0!"):
            Figure("квадрат", 0)
            
        with self.assertRaises(AssertionError):
            Figure("квадрат", -5)
            
        with self.assertRaisesRegex(AssertionError, "Невідомий тип фігури"):
            Figure("коло", 10)

    def test_all_figure_types_with_subtest(self):
        for fig_type in ["квадрат", "прямокутник", "трикутник"]:
            with self.subTest(figure=fig_type):
                fig = Figure(fig_type, 10)
                self.assertEqual(fig.get_figure_type, fig_type)

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

```python
import unittest
from unittest.mock import patch

def get_user_name():
    name = input("Введіть ваше ім'я: ")
    return f"Привіт, {name}!"

class TestUserInput(unittest.TestCase):
    
    @patch('builtins.input', return_value='Олексій')
    def test_get_user_name(self, mock_input):
        result = get_user_name()
        
        self.assertEqual(result, "Привіт, Олексій!")
        
        mock_input.assert_called_once_with("Введіть ваше ім'я: ")

if __name__ == '__main__':
    unittest.main(argv=['first-arg-is-ignored'], exit=False)
```
